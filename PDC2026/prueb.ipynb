```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Cargar imagen
ruta = "imagenes/colores.jpg"
imagen = cv2.imread(ruta)

if imagen is None:
    print("No se pudo cargar la imagen")
    exit()

# Convertir de BGR a RGB
imagen_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)

# Convertir a HSV
hsv = cv2.cvtColor(imagen, cv2.COLOR_BGR2HSV)

# Detectar color azul (cielo despejado)
azul_bajo = np.array([90, 50, 50])
azul_alto = np.array([130, 255, 255])

mascara_azul = cv2.inRange(hsv, azul_bajo, azul_alto)

# Detectar zonas grises/blancas (nubes)
gris_bajo = np.array([0, 0, 80])
gris_alto = np.array([180, 50, 255])

mascara_nubes = cv2.inRange(hsv, gris_bajo, gris_alto)

# Calcular porcentajes
pixeles_totales = imagen.shape[0] * imagen.shape[1]

pixeles_azules = cv2.countNonZero(mascara_azul)
pixeles_nubes = cv2.countNonZero(mascara_nubes)

porcentaje_azul = (pixeles_azules / pixeles_totales) * 100
porcentaje_nubes = (pixeles_nubes / pixeles_totales) * 100

# Clasificación del cielo
if porcentaje_azul > 50:
    estado = "Despejado"
    lluvia = "10%"
elif porcentaje_nubes > 50:
    estado = "Nublado"
    lluvia = "80%"
else:
    estado = "Parcialmente nublado"
    lluvia = "40%"

# Mostrar resultados
print("Estado del cielo:", estado)
print("Probabilidad de lluvia:", lluvia)

# Mostrar imágenes
plt.figure(figsize=(10,5))

plt.subplot(1,3,1)
plt.imshow(imagen_rgb)
plt.title("Imagen original")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(mascara_azul, cmap='Blues')
plt.title("Detección azul")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(mascara_nubes, cmap='gray')
plt.title("Detección nubes")
plt.axis("off")

plt.show()
```
