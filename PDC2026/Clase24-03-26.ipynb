Solarizacion 


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def solarizar(imagen, umbral):
    #aplicamos sol
    img_sol=np.where(imagen<umbral, imagen, 255-imagen)
    return img_sol

imagen= cv2.imread('imagen1.jpg')
img_gray= cv2.cvtColor(imagen,cv2.COLOR_BGR2GRAY)
#----
umbral=127
img_fin=solarizar(img_gray,umbral)
plt.figure(figsize=(12,6))
plt.subplot(121)
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))
plt.title('Original')
plt.subplot(122)
plt.imshow(img_fin,cmap='gray')
plt.title('Solarizada')
plt.show()
```

posterizar

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def posterizar(imagen,niveles):
    factor_escala=255/niveles
    img_pos=np.round(imagen/factor_escala)*factor_escala
    return img_pos.astype(np.uint8)
imagen= cv2.imread('imagen1.jpg')
img_gray= cv2.cvtColor(imagen,cv2.COLOR_BGR2GRAY)
nvl=4
img_fin=posterizar(img_gray,nvl)
plt.figure(figsize=(12,6))
plt.subplot(121)
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))
plt.title('Original')
plt.subplot(122)
plt.imshow(img_fin,cmap='gray')
plt.title('Posterizada')
plt.show()
```

posterizar y solarizar con colores

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def posterizar(imagen, niveles):
    factor_escala = 255 / niveles
    img_pos = np.round(imagen / factor_escala) * factor_escala
    return img_pos.astype(np.uint8)

def solarizar(imagen, umbral):
    img_sol = np.where(imagen < umbral, imagen, 255 - imagen)
    return img_sol.astype(np.uint8)

imagen = cv2.imread('imagen1.jpg')

nvl = 4
umbral = 128

img_posterizada = posterizar(imagen, nvl)
img_solarizada = solarizar(imagen, umbral)

plt.figure(figsize=(15,5))

plt.subplot(131)
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))
plt.title('Original')
plt.axis('off')

plt.subplot(132)
plt.imshow(cv2.cvtColor(img_posterizada, cv2.COLOR_BGR2RGB))
plt.title('Posterizada')
plt.axis('off')

plt.subplot(133)
plt.imshow(cv2.cvtColor(img_solarizada, cv2.COLOR_BGR2RGB))
plt.title('Solarizada')
plt.axis('off')

plt.show()
```

Falso color

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def falso_color(img_gray):
    escal_img=cv2.normalize(img_gray, None, 0, 255, cv2.NORM_MINMAX)
    falso_img=cv2.applyColorMap(escal_img, cv2.COLORMAP_JET)
    return falso_img
imagen= cv2.imread('imagen1.jpg')
img_gray= cv2.cvtColor(imagen,cv2.COLOR_BGR2GRAY)
img_fin= falso_color(img_gray)
plt.figure(figsize=(12,6))
plt.subplot(121)
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))
plt.title('Original')
plt.subplot(122)
plt.imshow(cv2.cvtColor(img_fin,cv2.COLOR_BGR2RGB))
plt.title('Falso Color')
plt.show()

```
