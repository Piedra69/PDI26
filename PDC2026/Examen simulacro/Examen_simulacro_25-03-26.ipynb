```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

alto = 256
ancho = 3

degradado = np.linspace(0, 255, alto, dtype=np.uint8)
imagen = np.tile(degradado[:, np.newaxis], (1, ancho))

imagen_redimensionada = cv2.resize(imagen, (300, 256))
imagen_Deg = cv2.applyColorMap(imagen_redimensionada, cv2.COLORMAP_VIRIDIS)

plt.figure(figsize=(8, 5))
plt.imshow(cv2.cvtColor(imagen_Deg, cv2.COLOR_BGR2RGB))
plt.title("Imagen degradado lineal")


cv2.imwrite("imagen_Deg.jpg", imagen_Deg)

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("imagen_Deg.jpg")

if img is None:
    print("No se pudo cargar imagen_Deg.jpg")
else:
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    gris = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    niveles = 4
    paso = 256 // niveles
    poster = (gris // paso) * paso
    poster = poster.astype(np.uint8)

    umbral = 128
    solar = np.where(gris < umbral, gris, 255 - gris)
    solar = solar.astype(np.uint8)

    plt.figure(figsize=(6, 5))
    plt.imshow(img_rgb)
    plt.title("Imagen original")
    plt.show()

    plt.figure(figsize=(6, 5))
    plt.imshow(poster, cmap="gray")
    plt.title("Posterizacion")
    plt.show()

    plt.figure(figsize=(6, 5))
    plt.imshow(solar, cmap="gray")
    plt.title("Solarizacion")
    plt.show()

    mapas = {
        "autumn": cv2.COLORMAP_AUTUMN,
        "bone": cv2.COLORMAP_BONE,
        "jet": cv2.COLORMAP_JET,
        "winter": cv2.COLORMAP_WINTER,
        "rainbow": cv2.COLORMAP_RAINBOW,
        "ocean": cv2.COLORMAP_OCEAN,
        "summer": cv2.COLORMAP_SUMMER,
        "spring": cv2.COLORMAP_SPRING,
        "cool": cv2.COLORMAP_COOL,
        "hsv": cv2.COLORMAP_HSV,
        "pink": cv2.COLORMAP_PINK,
        "hot": cv2.COLORMAP_HOT,
        "parula": cv2.COLORMAP_PARULA,
        "magma": cv2.COLORMAP_MAGMA,
        "inferno": cv2.COLORMAP_INFERNO,
        "plasma": cv2.COLORMAP_PLASMA,
        "viridis": cv2.COLORMAP_VIRIDIS,
        "cividis": cv2.COLORMAP_CIVIDIS,
        "twilight": cv2.COLORMAP_TWILIGHT,
        "twilight_shifted": cv2.COLORMAP_TWILIGHT_SHIFTED,
        "turbo": cv2.COLORMAP_TURBO,
        "deepgreen": cv2.COLORMAP_DEEPGREEN
    }

    n = len(mapas)
    cols = 4
    rows = (n + cols - 1) // cols

    fig1, axes1 = plt.subplots(rows, cols, figsize=(16, 4 * rows))
    axes1 = np.array(axes1).reshape(-1)

    for ax, (nombre, mapa) in zip(axes1, mapas.items()):
        img_color = cv2.applyColorMap(poster, mapa)
        img_rgb_color = cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB)
        ax.imshow(img_rgb_color)
        ax.set_title(f"Poster {nombre}")

    for ax in axes1[len(mapas):]:
        ax.axis("off")

    plt.tight_layout()
    plt.show()

    fig2, axes2 = plt.subplots(rows, cols, figsize=(16, 4 * rows))
    axes2 = np.array(axes2).reshape(-1)

    for ax, (nombre, mapa) in zip(axes2, mapas.items()):
        img_color = cv2.applyColorMap(solar, mapa)
        img_rgb_color = cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB)
        ax.imshow(img_rgb_color)
        ax.set_title(f"Solar {nombre}")

    for ax in axes2[len(mapas):]:
        ax.axis("off")

    plt.tight_layout()
    plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['figure.figsize'] = [8,8]

size_imagen = 20

A = np.ones((size_imagen, size_imagen)) * 255
N = np.zeros((size_imagen, size_imagen))

img_blanco = np.dstack((A, A, A))
img_negro = np.dstack((N, N, N))
img_amarillo = np.dstack((A, A, N))
img_rojo = np.dstack((A, N, N))

img_blanco = img_blanco.astype(np.uint8)
img_negro = img_negro.astype(np.uint8)
img_amarillo = img_amarillo.astype(np.uint8)
img_rojo = img_rojo.astype(np.uint8)

fila1  = np.hstack((img_blanco, img_blanco, img_amarillo, img_amarillo, img_amarillo, img_amarillo, img_blanco, img_blanco))
fila2  = np.hstack((img_blanco, img_amarillo, img_amarillo, img_amarillo, img_amarillo, img_amarillo, img_amarillo, img_blanco))
fila3  = np.hstack((img_amarillo, img_amarillo, img_negro, img_amarillo, img_amarillo, img_negro, img_amarillo, img_amarillo))
fila4  = np.hstack((img_amarillo, img_amarillo, img_amarillo, img_amarillo, img_amarillo, img_amarillo, img_amarillo, img_amarillo))
fila5  = np.hstack((img_amarillo, img_amarillo, img_amarillo, img_negro, img_negro, img_amarillo, img_amarillo, img_amarillo))
fila6  = np.hstack((img_amarillo, img_amarillo, img_rojo, img_amarillo, img_amarillo, img_rojo, img_amarillo, img_amarillo))
fila7  = np.hstack((img_blanco, img_amarillo, img_amarillo, img_rojo, img_rojo, img_amarillo, img_amarillo, img_blanco))
fila8  = np.hstack((img_blanco, img_blanco, img_amarillo, img_amarillo, img_amarillo, img_amarillo, img_blanco, img_blanco))

cara_feliz = np.vstack((fila1, fila2, fila3, fila4, fila5, fila6, fila7, fila8))
cara_feliz = cara_feliz.astype(np.uint8)

cara_feliz = cv2.resize(cara_feliz, (800, 800), interpolation=cv2.INTER_NEAREST)

plt.figure(figsize=(6,6))
plt.imshow(cara_feliz)
plt.title("Imagen original")
plt.axis("off")
plt.show()

cv2.imwrite("imagen_original.png", cv2.cvtColor(cara_feliz, cv2.COLOR_RGB2BGR))

img_bgr = cv2.cvtColor(cara_feliz, cv2.COLOR_RGB2BGR)

def mosaico_bayer(img, patron):
    b, g, r = cv2.split(img)
    mosaico = np.zeros((img.shape[0], img.shape[1]), dtype=np.uint8)

    if patron == "BG":
        mosaico[0::2, 0::2] = b[0::2, 0::2]
        mosaico[0::2, 1::2] = g[0::2, 1::2]
        mosaico[1::2, 0::2] = g[1::2, 0::2]
        mosaico[1::2, 1::2] = r[1::2, 1::2]

    elif patron == "GB":
        mosaico[0::2, 0::2] = g[0::2, 0::2]
        mosaico[0::2, 1::2] = b[0::2, 1::2]
        mosaico[1::2, 0::2] = r[1::2, 0::2]
        mosaico[1::2, 1::2] = g[1::2, 1::2]

    elif patron == "RG":
        mosaico[0::2, 0::2] = r[0::2, 0::2]
        mosaico[0::2, 1::2] = g[0::2, 1::2]
        mosaico[1::2, 0::2] = g[1::2, 0::2]
        mosaico[1::2, 1::2] = b[1::2, 1::2]

    elif patron == "GR":
        mosaico[0::2, 0::2] = g[0::2, 0::2]
        mosaico[0::2, 1::2] = r[0::2, 1::2]
        mosaico[1::2, 0::2] = b[1::2, 0::2]
        mosaico[1::2, 1::2] = g[1::2, 1::2]

    return mosaico

codigos = {
    "BG": cv2.COLOR_BAYER_BG2BGR,
    "GB": cv2.COLOR_BAYER_GB2BGR,
    "RG": cv2.COLOR_BAYER_RG2BGR,
    "GR": cv2.COLOR_BAYER_GR2BGR
}

for patron, codigo in codigos.items():
    mosaico = mosaico_bayer(img_bgr, patron)
    recuperada = cv2.cvtColor(mosaico, codigo)
    recuperada_rgb = cv2.cvtColor(recuperada, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(6,6))
    plt.imshow(recuperada_rgb)
    plt.title(f"Imagen recuperada Bayer {patron}")
    plt.axis("off")
    plt.show()

    cv2.imwrite(f"cara_bayer_{patron}.png", recuperada)
```
