```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("imagen1.jpg")
imgs = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gris = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

kernel_suavizado = np.ones((5, 5), np.float32) / 25
kernel_enfoque = np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]])
kernel_desenfoque = np.ones((3, 3), np.float32) / 9
kernel_realce = np.array([[0, 0, 0], [-1, 1, 0], [0, 0, 0]])
kernel_repujado = np.array([[-2, -1, 0], [-1, 1, 1], [0, 1, 2]])
kernel_bordes = np.array([[0, 1, 0], [1, -4, 1], [0, 1, 0]])
kernel_sobel = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]])
kernel_sharpen = np.array([[1, -2, 1], [-2, 5, -2], [1, -2, 1]])
kernel_norte = np.array([[1, 1, 1], [1, -2, 1], [-1, -1, -1]])
kernel_oeste = np.array([[-1, 1, 1], [-1, -2, 1], [-1, 1, 1]])

kernel_gauss = np.array([
    [1, 2, 3, 2, 1],
    [2, 7, 11, 7, 2],
    [3, 11, 17, 11, 3],
    [2, 7, 11, 7, 2],
    [1, 2, 3, 2, 1]
], np.float32)
kernel_gauss = kernel_gauss / np.sum(kernel_gauss)

filtros = [
    ("Original", imgs, gris),
    ("Suavizada 5x5", cv2.filter2D(imgs, -1, kernel_suavizado), cv2.filter2D(gris, -1, kernel_suavizado)),
    ("Enfoque", cv2.filter2D(imgs, -1, kernel_enfoque), cv2.filter2D(gris, -1, kernel_enfoque)),
    ("Desenfoque", cv2.filter2D(imgs, -1, kernel_desenfoque), cv2.filter2D(gris, -1, kernel_desenfoque)),
    ("Realce de bordes", cv2.filter2D(imgs, -1, kernel_realce), cv2.filter2D(gris, -1, kernel_realce)),
    ("Repujado", cv2.filter2D(imgs, -1, kernel_repujado), cv2.filter2D(gris, -1, kernel_repujado)),
    ("Detección de bordes", cv2.filter2D(imgs, -1, kernel_bordes), cv2.filter2D(gris, -1, kernel_bordes)),
    ("Sobel", cv2.filter2D(imgs, -1, kernel_sobel), cv2.filter2D(gris, -1, kernel_sobel)),
    ("Sharpen", cv2.filter2D(imgs, -1, kernel_sharpen), cv2.filter2D(gris, -1, kernel_sharpen)),
    ("Filtro Norte", cv2.filter2D(imgs, -1, kernel_norte), cv2.filter2D(gris, -1, kernel_norte)),
    ("Filtro Oeste", cv2.filter2D(imgs, -1, kernel_oeste), cv2.filter2D(gris, -1, kernel_oeste)),
    ("Filtro Gauss", cv2.filter2D(imgs, -1, kernel_gauss), cv2.filter2D(gris, -1, kernel_gauss))
]

plt.figure(figsize=(28, 72))

for i, (titulo, color, gray) in enumerate(filtros):
    plt.subplot(len(filtros), 2, i * 2 + 1)
    plt.imshow(color)
    plt.title(titulo + " - Color")
   

    plt.subplot(len(filtros), 2, i * 2 + 2)
    plt.imshow(gray, cmap="gray")
    plt.title(titulo + " - Gris")
   

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convolucion_media_h(imagen, tkernel):
    kernel = np.ones((1, tkernel), np.float32) / tkernel
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

def convolucion_media_v(imagen, tkernel):
    kernel = np.ones((tkernel, 1), np.float32) / tkernel
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

def convolucion_media_diagonal(imagen, tkernel):
    kernel = np.zeros((tkernel, tkernel), np.float32)

    for i in range(tkernel):
        kernel[i, i] = 1

    kernel = kernel / tkernel
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

def convolucion_media_diagonal_inversa(imagen, tkernel):
    kernel = np.zeros((tkernel, tkernel), np.float32)

    for i in range(tkernel):
        kernel[i, tkernel - 1 - i] = 1

    kernel = kernel / tkernel
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c


img = cv2.imread("imagen1.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

tkernel = 35

horizontal = convolucion_media_h(img, tkernel)
vertical = convolucion_media_v(img, tkernel)
diagonal = convolucion_media_diagonal(img, tkernel)
diagonal_inversa = convolucion_media_diagonal_inversa(img, tkernel)

plt.figure(figsize=(12, 9))

plt.subplot(3, 2, 1)
plt.imshow(vertical)
plt.title("Convolucionada Vertical")
plt.axis("off")

plt.subplot(3, 2, 2)
plt.imshow(horizontal)
plt.title("Convolucionada Horizontal")
plt.axis("off")

plt.subplot(3, 2, 3)
plt.imshow(diagonal_inversa)
plt.title("Convolucionada Diagonal Inversa")
plt.axis("off")

plt.subplot(3, 2, 4)
plt.imshow(diagonal)
plt.title("Convolucionada Diagonal")
plt.axis("off")

plt.subplot(3, 2, 6)
plt.imshow(img)
plt.title("Original de Nuevo")
plt.axis("off")

plt.tight_layout()
plt.show()
```
