Serparar por canales sin coloreado


```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['figure.figsize'] = [25,3]

img = cv2.imread('colores.jpg', cv2.IMREAD_COLOR)
imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

r = img[:,:,2]
g = img[:,:,1]
b = img[:,:,0]

r_img = np.zeros_like(imgRGB)
g_img = np.zeros_like(imgRGB)
b_img = np.zeros_like(imgRGB)

r_img[:,:,0] = r
g_img[:,:,1] = g
b_img[:,:,2] = b

plt.subplot(141)
plt.imshow(imgRGB)
plt.title('Original')
plt.axis('off')

plt.subplot(142)
plt.imshow(r_img)
plt.title('Rojo')
plt.axis('off')

plt.subplot(143)
plt.imshow(g_img)
plt.title('Verde')
plt.axis('off')

plt.subplot(144)
plt.imshow(b_img)
plt.title('Azul')
plt.axis('off')

plt.show()
```

```python
#imagen 1
import cv2
import matplotlib.pyplot as plt


img1_bgr = cv2.imread("imagen1.jpg")

if img1_bgr is None:
    print("Error: no se pudo cargar imagen1.jpg")
else:

    def mostrar_imagen_rgb(img_bgr, titulo):
        img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
        plt.figure(figsize=(6, 6))
        plt.title(titulo)
        plt.imshow(img_rgb)
        plt.show()

    def mostrar_imagen_gris(img_gray, titulo):
        plt.figure(figsize=(6, 6))
        plt.title(titulo)
        plt.imshow(img_gray, cmap="gray")
        plt.show()

    
    mostrar_imagen_rgb(img1_bgr, "Imagen 1 (BGR leída, mostrada como RGB)")

  
    img1_gray = cv2.cvtColor(img1_bgr, cv2.COLOR_BGR2GRAY)

    cv2.imwrite("imagen1_gris.jpg", img1_gray)

    cv2.imwrite("imagen1_bgr.jpg", img1_bgr)


    mostrar_imagen_gris(img1_gray, "Imagen 1 (Escala de grises)")

   
 
    canal_b1 = img1_bgr[:, :, 0]
    canal_g1 = img1_bgr[:, :, 1]
    canal_r1 = img1_bgr[:, :, 2]

    falso1_bgr = cv2.applyColorMap(img1_gray, cv2.COLORMAP_JET)
    falso1_rgb = cv2.cvtColor(falso1_bgr, cv2.COLOR_BGR2RGB)

  
    img1_rgb = cv2.cvtColor(img1_bgr, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(12, 8))
    plt.suptitle("Imagen 1 - Subplot Modo 1 (plt.subplot)")

    plt.subplot(2, 3, 1); plt.title("Color (RGB)"); plt.imshow(img1_rgb); 
    plt.subplot(2, 3, 2); plt.title("Grises"); plt.imshow(img1_gray, cmap="gray"); 
    plt.subplot(2, 3, 3); plt.title("Color falso (JET)"); plt.imshow(falso1_rgb); 
    plt.subplot(2, 3, 4); plt.title("Canal B (gris)"); plt.imshow(canal_b1, cmap="gray"); 
    plt.subplot(2, 3, 5); plt.title("Canal G (gris)"); plt.imshow(canal_g1, cmap="gray"); 
    plt.subplot(2, 3, 6); plt.title("Canal R (gris)"); plt.imshow(canal_r1, cmap="gray"); 

    plt.show()

  
    fig, ax = plt.subplots(2, 3, figsize=(12, 8))
    fig.suptitle("Imagen 1 - Subplot Modo 2 (plt.subplots)")

    ax[0, 0].set_title("Color (RGB)"); ax[0, 0].imshow(img1_rgb); ax[0, 0]
    ax[0, 1].set_title("Grises"); ax[0, 1].imshow(img1_gray, cmap="gray"); ax[0, 1]
    ax[0, 2].set_title("Color falso (JET)"); ax[0, 2].imshow(falso1_rgb); ax[0, 2]
    ax[1, 0].set_title("Canal B"); ax[1, 0].imshow(canal_b1, cmap="gray"); ax[1, 0]
    ax[1, 1].set_title("Canal G"); ax[1, 1].imshow(canal_g1, cmap="gray"); ax[1, 1]
    ax[1, 2].set_title("Canal R"); ax[1, 2].imshow(canal_r1, cmap="gray"); ax[1, 2]

    plt.tight_layout()
    plt.show()
```
