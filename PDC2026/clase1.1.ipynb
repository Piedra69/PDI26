```python
import cv2
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize']=[25,3]
img=cv2.imread('colores.jpg',cv2.IMREAD_COLOR)
#DIVISION POR CANALES
plt.subplot(141)
plt.title('imagen original')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))

plt.subplot(142)
plt.title('CANAL ROJO')
plt.imshow(img[:,:,0], cmap='gray')

plt.subplot(143)
plt.title('CANAL VERDE')
plt.imshow(img[:,:,1], cmap='gray')

plt.subplot(144)
plt.title('CANAL AZUL')
plt.imshow(img[:,:,2], cmap='gray')

plt.show()
```

Serparar por canales sin coloreado


```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['figure.figsize'] = [25,3]

img = cv2.imread('colores.jpg', cv2.IMREAD_COLOR)
imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

r = img[:,:,2]
g = img[:,:,1]
b = img[:,:,0]

r_img = np.zeros_like(imgRGB)
g_img = np.zeros_like(imgRGB)
b_img = np.zeros_like(imgRGB)

r_img[:,:,0] = r
g_img[:,:,1] = g
b_img[:,:,2] = b

plt.subplot(141)
plt.imshow(imgRGB)
plt.title('Original')
plt.axis('off')

plt.subplot(142)
plt.imshow(r_img)
plt.title('Rojo')
plt.axis('off')

plt.subplot(143)
plt.imshow(g_img)
plt.title('Verde')
plt.axis('off')

plt.subplot(144)
plt.imshow(b_img)
plt.title('Azul')
plt.axis('off')

plt.show()
```
